```python
### Statistical Testing.............
```

```python
#### OverTime vs Attrition.............

from scipy.stats import chi2_contingency

# Contingency table banao
contingency_table = pd.crosstab(df['OverTime'], df['Attrition'])
print(contingency_table)

chi2, p, dof, expected = chi2_contingency(contingency_table)
print(f"Chi-square statistic: {chi2:.2f}")
print(f"P-value: {p:.5f}")

if p < 0.05:
    print("Result: Significant relationship (OverTime affects Attrition)")
else:
    print("Result: No significant relationship")
```

```python
### Department vs Attrition.....

contingency_table2 = pd.crosstab(df['Department'], df['Attrition'])
chi2, p, dof, expected = chi2_contingency(contingency_table2)
print(f"P-value: {p:.5f}")
```

```python
###MonthlyIncome:.....
from scipy.stats import ttest_ind

income_yes = df[df['Attrition']=='Yes']['MonthlyIncome']
income_no = df[df['Attrition']=='No']['MonthlyIncome']

t_stat, p_value = ttest_ind(income_yes, income_no)
print(f"T-statistic: {t_stat:.2f}")
print(f"P-value: {p_value:.5f}")

if p_value < 0.05:
    print("Result: Significant difference in income between groups")
```

```python
import sqlite3

conn = sqlite3.connect('hr_data.db')

df.to_sql('employees', conn, if_exists='replace', index=False)

print("Data successfully loaded into SQL database!")
```

```python
query = "SELECT * FROM employees LIMIT 5;"
result = pd.read_sql(query, conn)
result
```

```python
query1 = """
SELECT Department,
       COUNT(*) as Total_Employees,
       SUM(CASE WHEN Attrition='Yes' THEN 1 ELSE 0 END) as Attritions,
       ROUND(SUM(CASE WHEN Attrition='Yes' THEN 1.0 ELSE 0 END) / COUNT(*) * 100, 2) as Attrition_Rate
FROM employees
GROUP BY Department
ORDER BY Attrition_Rate DESC;
"""
pd.read_sql(query1, conn)
```

```python
query2 = """
SELECT Department,
       AVG(MonthlyIncome) as Avg_Income,
       RANK() OVER (ORDER BY AVG(MonthlyIncome) DESC) as Income_Rank
FROM employees
GROUP BY Department;
"""
pd.read_sql(query2, conn)
```

```python
query3 = """
WITH high_risk AS (
    SELECT *
    FROM employees
    WHERE OverTime = 'Yes' AND JobSatisfaction <= 2
)
SELECT Department, JobRole, COUNT(*) as High_Risk_Count
FROM high_risk
GROUP BY Department, JobRole
ORDER BY High_Risk_Count DESC;
"""
pd.read_sql(query3, conn)
```
